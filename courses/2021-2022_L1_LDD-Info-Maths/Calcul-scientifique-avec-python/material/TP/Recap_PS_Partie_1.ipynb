```python
import math
import numpy as np

def Benford_cutoff(f):
    mag=10.**math.floor(math.log(f)/math.log(10))
    return math.floor(f/mag)
print(Benford_cutoff(0.37))
print(Benford_cutoff(9356.839))
```

```python
def measure_time(func, N, *args):
    start_time = time.time()
    for _ in range(N):
        func(*args)
    end_time = time.time()
    return end_time - start_time
```
