# Exercice 3
Donnez une fonction permettant de tester si un graphe est bipartite ou non. On utilisera pour cela un parcours en profondeur que l'on modifiera.

Voici l'algorithme du parcours en profondeur basique. 

```python
def Profondeur_pile(G,s):
    dvu={}
    for i in G.vertices():
        dvu[i]=False   
    P = [s]
    print(s)
    dvu[s] = True
    while len(P)!=0:
        som = P[-1]
        trouve = False
        for i in G.neighbors(som):
            if trouve==False:
                if dvu[i] == False:
                    P.append(i)
                    print(i)
                    dvu[i] = True
                    trouve = True
        if trouve == False:
            P.pop()
```
