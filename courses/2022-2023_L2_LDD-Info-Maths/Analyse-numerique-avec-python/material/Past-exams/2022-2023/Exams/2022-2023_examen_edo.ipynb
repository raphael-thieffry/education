On se propose dans cette partie d'investir dans la résolution numérique 
de l'équation différentielle ordinaire : 
\begin{equation}
(EDO)\quad \left\{
\begin{array}{rcl}
x'(t)  &=& f(t, x(t)), \quad t \in ]t_0, t_0 + T[,\\
x(t_0) &=& x^0
\end{array}
\right.
\end{equation}
où  $f \in C^r ([t_0, t_0 + T]\times \mathbb{R}; \mathbb{R}), x^0 \in \mathbb{R}$ et où $T$ est un réel strictement positif et $t_0\in \mathbb{R}$ (les valeurs de $r, t_0, T, x^0$ seront précisées dans chaque exercice).

Pour résoudre numériquement ce problème de Cauchy, on se donne un entier $N \in \mathbb{N}^*$, 
et on considère une subdivision uniforme $t_0 < t_1 < \ldots < t_N = (t_0 + T)$ de $[t_0, t_0 + T]$
de pas $\Delta t = \frac{T}{N}$, définie par $t_n = t_0 + n\, \Delta t \quad (0\leq n \leq N)$. 
Puis on construit la suite $(x_n)_{0\leq n\leq N}$ telle que $x_n$ soit une valeur approchée de $x(t_n)$ pour tout $ n \in \{0, \ldots, N\}$, via un schéma numérique **explicite**. 

**Remarque:**
Le cahier dse charges nous demande de bâtir un schéma numérique pour ce problème 
>- en tirant partie du fait que 
(quelque soit $N$ ou $\Delta t$  et $x_0$ que l'on choisira) on pourra mettre à notre disposition (une bonne) approximation de $x_1$ et ceci indépendamment de notre schéma. 
>- Il sera alors question pour nous de préciser les contraintes sur la valeur de $x_1$ à fournir. Et vérifier si celle fournie les satisfait.

## Exercice 2 : Mise en oeuvre ( A faire sur ce notebook)

Pour mettre en oeuvre le schéma, on aura besoin d'une fonction qui permette de construitre $x_1$,
à partir de $x_0$ cette fonction sera appellée `demarrage`

Elle est fournie comme précisée dans le cahier des charges:
 

```python
def demarrage(f, t0, x0, dt):
    '''
    Fonction permettant de construire la donnée initiale commplémentaire x1
    Entrees:
    --------
        f  :  second memebre de l'edo 
        t0 :  instant initial
        x0 :  aleur initiale de la solution
        dt :  le pas de temps
    Sortie:
        t1 :  instant qui suit l'instant t0
        x1 :  la solution à l'instant t1
    ''' 
    t1 = t0 + dt
    f0 = f(t0,x0)
    f1 = f(t0 + dt,  x0 + dt * f0)
    x1 = x0 +  dt * ( f0  +  f1 ) / 2
    return t1, x1
```

```python
def analyse_erreur_schema(f, t0, tf, x_exacte, monSchema, title):
    dts = (tf - t0)/(2.**np.arange(3,11))
    E = 0 * dts
    for i, dt in enumerate(dts):
        N = int( (tf - t0)/dt)
        t_, x_ = monSchema(f, t0, tf, x_exacte(t0), N)
        xe = x_exacte(t_)
        E[i] = np.max(np.abs(xe - x_))
    fmt1 = "| {:^20s} | {:^20s} | {:^23s} |".format('dt', 'E_dt', "log(E_dt/E_{dt/2})/log(2)") 
    print('-'* len(fmt1))
    print(fmt1)
    print('-'* len(fmt1))
    fmt3 = "| {0:^20.2E} | {1:^20.4E} | {2:^23s} |"
    fmt2 = "| {0:^20.2E} | {1:^20.4E} | {2:^23.2f} |"
    print(fmt3.format(dts[0], E[0],'-'))
    for i in range(1, dts.size):
        print(fmt2.format(dts[i], E[i], np.log(E[i-1]/E[i])/np.log(2.) ))
    print('-'* len(fmt1))
    print("| {} \n| Estimation de l'ordre du Schéma {} pour EDO".format(title, monSchema.__name__))
    print('-'* len(fmt1))
```
