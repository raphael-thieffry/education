**Q1-2-1: Réduction de l'intervalle de recherche au quart, par dichotomie**

 - Soit la fonction g(x) = f(x) - x. En appliquant seulement deux étapes de l'algorithme de dichotomie, déterminer un nouel intervalle [a,b] de longueur le quart du précédent dans lequel on est certain de trouver le point fixe de $f$. 
 
 - Reprendre le graphique ci-dessus (de la **Question 1-1**) pour ce nouvel intervalle.
 
 (*pour éviter les problèmes d'échelles, on ne représentera plus les axes des coordonnées*)
 


```python
# UTILISEZ CETTE CELLULE POUR A QUESTION 1-2-1.

def g(x): return f(x) - x


#intervalle de départ
a, b = -1., 0.5
long_initial = b - a
print("Départ [a, b] = [{:}, {:}] lonngueur = {:}".format(a,b, b-a)) 

#on applique dichotomie 1 fois
c = (a + b)/2
if g(a) * g(c)  < 0 :
    a, b = a, c
else:
    a,b = c,b
#on applique dichotomie une seconde fois    
c = (a + b)/2
if g(a) * g(c)  < 0 :
    a, b = a, c
else:
    a,b = c,b    
    
# on stocke cette valeur de a, b
a1, b1 = a, b

long_final = b1 - a1

print("Final  [a, b] = [{:}, {:}] lonngueur = {:}".format(a1,b1, b1-a1))   
print("On bien : longueur_initial/longeueur_final = {:}".format(long_initial /long_final))

```

**Q1-2-2: Déduction de la valeur approchée du point fixe**

Dans le graphique précédant, l'analyste observe graphiquement que sur ce nouvel intervalle la courbe de $f$ est presqu' un segment (i.e une droite). It décide donc d'utiliser une approximation affine de la fonction pour estimer le point fixe. Pour cela il cherche une approximation de f(x)  sur cet intervalle sour la forme $f(x) \approx \alpha (x - a) + \beta$.

- Déterminer $\alpha, \beta $ en fonction de f(a), f(b) puis déduire une approximation du point fixe $x^*$.

- Reprendre alors le graphique et l'intervalle de la question **Q1-1** en y ajoutant cette fois le point fixe avec une étoile de couleur verte.  

```python
# UTILISEZ CETTE CELLULE POUR A QUESTION 1-2-2.

alpha = (f(b1) - f(a1))/(b1 - a1)
beta  =  f(a)
xstar =  (beta - a1 * alpha)/(1 - alpha)   # valeur approchée du point fixe x^*

# attention le graphique doit être sur [a, b] = [-1., 0.5] de départ

a, b = -1., 0.5
x = np.linspace(a, b, 60)
y = f(x)

plt.plot(x,y, '-b', label="y = f(x)")
plt.plot(x,x, color='orange', label="y = x")
plt.plot(x,x*0, "--k")
plt.plot([0,0],[np.min([x,y]), np.max([x,y])] , "--k")
plt.scatter(xstar, xstar, color='green', label="point fixe")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.title("courbes sur [{:},{:}]".format(a,b))
plt.legend()
plt.show()
```

**Q1-2-3: Optimisation**

L'analyste remarque que le calcul de $\alpha$,$\beta$ nécessite deux évaluations de la fonction et une division. 
Puis en pointant le fait que la fonction $f$ et sa dérivéé sont identiques, il propose une expression de 
$\alpha$ et $\beta$ ne nécessitant qu'une seule évaluation de la fonction et aucune autre opération.
 
- Donner alors les expressions de $\alpha$ et $\beta$ qu'il a proposé 

- Et reprendre le graphique de la question **Q1-2-2** ci-dessus.
  

```python
# UTILISEZ CETTE CELLULE POUR A QUESTION 1-2-3.
alpha = f(a1)
beta  = alpha
xstar =  (beta - a1 * alpha)/(1 - alpha)   # valeur approchée du point fixe x^*

a, b = -1., 0.5
x = np.linspace(a, b, 60)
y = f(x)

plt.plot(x,y, '-b', label="y = f(x)")
plt.plot(x,x, color='orange', label="y = x")
plt.plot(x,x*0, "--k")
plt.plot([0,0],[np.min([x,y]), np.max([x,y])] , "--k")
plt.scatter(xstar, xstar, color='green', label="point fixe")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.title("courbes sur [{:},{:}]".format(a,b))
plt.legend()
plt.show()
```

### Question 2-2 : Mise en oeuvre
Fournir une fonction une `pointFixe` qui implémente la méthode du point fixe.

```python
def pointFixe(f, x0, tol=1e-6, iterMax=100):
    """
    Entrees:
         f   : fonction dont on cherche le point fixe
         x0  : le point de départ
         tol : la tolérance pour le test d'arrêt
         iterMax : le nombre maximum permis d'itérations 
     Sorties:
         x   : l'approximation du point fixe
         k   : le nombre d'itérations effectuées
    """
    conv = False
    x = x0
    k = 0
    while conv ==  False:
        #calcul
        x1 = f(x)
        #test convergence
        conv = np.abs(x1 - x) < tol or k > iterMax 
        #mise à jour
        x = x1
        k = k + 1
    return x, k
```

```python
# oubien comme en TP 
def pointFixe(f, x0, tol=1e-6, iterMax=100):
    x = x0
    niter = 0
    fx = f(x)
    while abs(fx-x) > tol and niter < iterMax:
        x = fx
        fx = f(x)
        niter += 1
    return x, niter
```
